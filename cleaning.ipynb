```python
#data load
import pandas as pd

df = pd.read_excel(r"C:\Users\saina\Downloads\sales_raw_500.xlsx")
```

```python
df
```

```python
#duplicate removes

df = df.drop_duplicates()
```

```python
df
```

```python
#fix date column
df['Order_Date'] = pd.to_datetime(df['Order_Date'],error='coerce')
```

```python
df = df.dropna(subset=['Order_Date'])
```

```python
df
```

```python
df.loc[:, 'Profit'] = df['Sales'] - df['Cost']
df.loc[:, 'Year'] = df['Order_Date'].dt.year
df.loc[:, 'Month'] = df['Order_Date'].dt.month
```

```python
df
```

```python
df.to_csv('sales_cleaned.csv',index = False)
```
